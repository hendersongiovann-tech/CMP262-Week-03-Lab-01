Creating and Accessing a list 

```python
courses = ["Algebra", "Biology", "Computer Science", "English", "History"]

print("Entire List - ", courses)
print("First Course - ", courses[0])
print("Third Course - ", courses[2])
print("Last Course - ", courses[-1])
print("Length of Courses - ", len(courses))

print("**First Three Courses** - ", courses[:3])
```

Modifying A List 

```python
favorite_foods = ["Spaghetti", "Pizza", "cheeseburger", "Fried Chicken", "Ice Cream"]

print("Original List - ", favorite_foods)

favorite_foods[4] = "Cheesecake"
favorite_foods.append("Meatloaf")
favorite_foods.remove("Pizza")

print("Updated List - ", favorite_foods)
```

Working with Numbers 

```python
scores = [14, 20, 35, 17, 40, 31]

print("Scores - ", scores)
print("Highest Score - ", max(scores))
print("Lowest Score - ",min(scores))
print("Sum Of All Scores - "sum(scores))
print("Length of Scores - ", len(scores))

print("The Average Score - ",sum(scores) / len(scores))
```

Looping Through a List 

```python
cities = ["New York", "Los Angelos", "Houston", "Newark", "Miami"]

for city in cities:
    print(city)

for city in cities:
    print("I would Like to Visit", (city))    
```

Student Scores 

```python
student_scores = [78, 92, 85, 67, 95, 88, 73]

print("Hight Student Score - ", max(student_scores))
print("Lowest Student Score - ", min(student_scores))
print("The Sum Of all the Student Scores - ",sum(student_scores) / len(student_scores))

for scores in student_scores:
    if scores >= 80:
        print("**Scores 80 or Higher** - ", scores)
```
